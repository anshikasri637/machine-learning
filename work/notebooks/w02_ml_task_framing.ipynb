## 1. My lane as an ML task (type)

**Type: unsupervised clustering** (specifically K-Means).

Unlike the scoring or prediction lanes, there's no target column to predict here — the goal is to discover natural groupings in the data, not forecast a labeled outcome.

## 2. Target or proxy

**There isn't one, by design.** Clustering is unsupervised: cluster membership *is* the output, not a prediction of a pre-existing label. This matters because it sidesteps one of the biggest traps in this internship — there's no product decision flag (`health_score`, `priority_score`, etc.) to accidentally leak in as a label, because there's no label at all.

## 3. Success metric

Three things, since clustering has no single accuracy number:
1. **Silhouette score** — how well-separated the clusters are (higher = better defined groups).
2. **Stability** — do cluster sizes and membership stay roughly consistent across different random seeds? If not, the clusters aren't reliable enough to act on.
3. **Human-readability** — can a reviewer look at a cluster's profile and immediately understand what kind of page it represents? A cluster that's statistically clean but meaningless to a human isn't useful here.

## 5. Why ML beats a fixed rule here

FlyRank's existing `impression_tier` is a fixed rule along a single axis (impression volume, cut into 4 quantile buckets). But a page's real situation depends on the *combination* of traffic, trend direction, engagement, staleness, and position at once — and those interact in ways that are too messy for a simple if/else rule to capture. For example, two pages with identical impressions can be in completely different situations if one is declining steeply with weak engagement and the other is flat with strong engagement. Clustering finds these joint patterns automatically instead of requiring someone to hand-write every combination of thresholds.
