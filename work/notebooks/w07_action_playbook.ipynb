## 2. Intended use and limits

**Who uses this:** a content reviewer deciding which pages to look at first, given limited weekly review capacity.

**Where it stops being valid:** this ranks *review priority*, not guaranteed outcomes. It cannot say a refresh will recover traffic — that needs a causal design (e.g. an A/B test), which this snapshot data can't provide. It also reflects one 90-day window; client history length varies, so a page's cluster could shift once more history accumulates.

## 3. Human review + the no-go list

**Must check before acting:**
- Confirm the page's decline isn't consolidation (a sibling page absorbing its traffic) or seasonality before recommending a rewrite.
- For cluster 0, investigate why `word_count` is missing (batch, provider, time period) before assigning any action at all.

**Never automate:** publishing a rewrite, merge, or prune action directly from cluster membership without a human sign-off — this is a review queue, not an autopilot.

## 4. Monitoring / retrain triggers
