## 2. The question: decision, action, cost of a wrong call

**Question:** What performance archetypes exist across FlyRank's content inventory, and what action should a reviewer take for each?

**Decision it improves:** which pages a content reviewer looks at first, and what they do (protect / improve / rewrite / merge / prune / monitor) — given limited review capacity.

**Who acts on it:** a human content reviewer with a fixed number of pages they can realistically check per week.

**Cost of a wrong call:** if a page is put in the wrong archetype bucket, the reviewer either wastes time on a page that won't respond to a refresh (false positive) or misses a page that's genuinely at risk (false negative). Because this is decision-support, not automation, the real safety net is that a human still reviews the top picks before acting.
