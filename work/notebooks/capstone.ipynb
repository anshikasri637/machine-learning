## 1. Question

What performance archetypes exist across FlyRank's content inventory, and what action should a reviewer take for each, given limited review capacity?

## 2. Data

`content_refresh_anonymized.csv` — 30,000 pseudonymized pages, 44 columns, 90-day trailing window. Excluded: keyword-only fields (`search_volume`, `competition`, `cpc`), redundant tier columns, and content-generation metadata (`provider_used`, `model_used`). No client names, URLs, or raw queries anywhere in this data.

## 3. Methodology

K-Means clustering (K=7, chosen via silhouette score after excluding features that were causing the clusters to just rediscover `content_type`). Numeric features log-transformed and standardized; two informative-null cases (`trend_pct`, `avg_position==0`) fixed based on audit findings rather than blindly imputed. Baseline: FlyRank's existing `impression_tier` (single-dimension rule). Validation: stability check across two random seeds.

## 4. Results (vs baseline)

The baseline sorts pages into 4 traffic-only buckets. The 7-cluster solution separates pages that share a traffic tier but differ sharply in trend and engagement — surfacing, for example, a distinct "steepest decline + weakest engagement + stalest" cluster that the baseline would have buried inside its "excellent" or "good" tier alongside healthy pages.

## 5. Limitations

- Decision-support only — no causal claim that any action will recover traffic.
- One cluster is confounded with a data-tracking gap (`word_count` missingness), not a clean behavioral signal, and is flagged rather than named as a true archetype.
- The smallest cluster shows more boundary movement across reseeding — reported honestly rather than smoothed over.
- Single 90-day snapshot; unbalanced tracking history across clients.

## 6. Ranked recommendations

See the full table in `w07_action_playbook.ipynb` — top priority is the steepest-decline/weakest-engagement cluster, followed by the high-value flagship-pages-in-decline cluster, with the two low-demand clusters left as monitor-only given their low ceiling.

## 7. Artifacts the paper embeds

Cluster profile table, elbow/silhouette chart, PCA scatter plot, and the ranked action table — all generated in `w05_model.ipynb` and `w07_action_playbook.ipynb`.
